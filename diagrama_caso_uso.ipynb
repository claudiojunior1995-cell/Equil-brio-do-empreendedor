# Diagrama de Caso de Uso (UML)
## Projeto: Equilíbrio Empreendedor

```python
# PASSO 1: Instalar dependências
!pip install plantuml -q
!apt-get install -y default-jre > /dev/null 2>&1
```

```python
# PASSO 2: Instalar e configurar graphviz (necessário para PlantUML)
!apt-get install -y graphviz > /dev/null 2>&1
print('✓ Dependências instaladas')
```

```python
# PASSO 3: Definir o diagrama UML em PlantUML

diagrama_uml = """
@startuml Diagrama_Caso_Uso_Equilibrio_Empreendedor
!define DIRECTION top to bottom
skinparam BackgroundColor #FFFFFF
skinparam ActorBackgroundColor #FFD700
skinparam UsecaseBackgroundColor #87CEEB
skinparam ArrowColor #333333
skinparam BorderColor #333333

title Diagrama de Caso de Uso - Equilíbrio Empreendedor

' === ATORES ===
actor "Usuário\n(Empreendedor)" as Usuario
actor "Amigos" as Amigos
actor "Sistema" as Sistema

' === CASOS DE USO: GERENCIAR ACESSO ===
rectangle "Gerenciar Acesso" {
  usecase "UC01: Efetuar\nCadastro/Login" as UC01
  usecase "UC02: Selecionar\nPersonagem/Avatar" as UC02
}

' === CASOS DE USO: OPERAR PARTIDA ===
rectangle "Operar Partida" {
  usecase "UC03: Iniciar\nPartida" as UC03
  usecase "UC04: Criar Sala/\nConvidar Amigos" as UC04
  usecase "UC05: Realizar\nJogada (Turno)" as UC05
}

' === CASOS DE USO: GESTÃO FINANCEIRA ===
rectangle "Gestão Financeira" {
  usecase "UC06: Gerenciar\nCarteira Empresa (70%)" as UC06
  usecase "UC07: Gerenciar\nCarteira Pessoal (30%)" as UC07
  usecase "UC08: Formalizar-se\ncomo MEI" as UC08
  usecase "UC09: Evitar Multa\nde Gestão" as UC09
}

' === CASOS DE USO: SUPORTE E RESULTADOS ===
rectangle "Suporte e Resultados" {
  usecase "UC10: Consultar\nChatbot (Guia)" as UC10
  usecase "UC11: Visualizar\nRanking Final" as UC11
}

' === ASSOCIAÇÕES: USUÁRIO ===
Usuario --> UC01
Usuario --> UC02
Usuario --> UC03
Usuario --> UC05
Usuario --> UC06
Usuario --> UC07
Usuario --> UC08
Usuario --> UC10
Usuario --> UC11

' === ASSOCIAÇÕES: AMIGOS ===
Amigos --> UC04
Amigos --> UC05

' === ASSOCIAÇÕES: SISTEMA ===
Sistema --> UC06
Sistema --> UC07
Sistema --> UC09

' === RELACIONAMENTOS: INCLUDES ===
UC03 ..> UC02 : include
UC04 ..> UC01 : include
UC05 ..> UC06 : include
UC05 ..> UC07 : include
UC05 ..> UC09 : include

' === RELACIONAMENTOS: EXTENDS ===
UC08 ..|> UC05 : extend
UC10 ..|> UC05 : extend

@enduml
"""

print('✓ Diagrama UML definido')
```

```python
# PASSO 4: Salvar o diagrama em arquivo .puml

with open('diagrama_caso_uso.puml', 'w', encoding='utf-8') as f:
    f.write(diagrama_uml)

print('✓ Arquivo diagrama_caso_uso.puml criado')
```

```python
# PASSO 5: Gerar imagem do diagrama usando PlantUML

import subprocess
import os

# Baixar o JAR do PlantUML
os.system('wget -q https://sourceforge.net/projects/plantuml/files/plantuml.jar/download -O plantuml.jar')

# Gerar a imagem PNG
result = subprocess.run(
    ['java', '-jar', 'plantuml.jar', '-Tpng', 'diagrama_caso_uso.puml'],
    capture_output=True,
    text=True
)

if result.returncode == 0:
    print('✓ Diagrama gerado com sucesso: diagrama_caso_uso.png')
else:
    print('⚠ Usando método alternativo...')
    os.system('java -jar plantuml.jar -Tpng diagrama_caso_uso.puml')
```

```python
# PASSO 6: Exibir o diagrama

from IPython.display import Image, display

if os.path.exists('diagrama_caso_uso.png'):
    display(Image('diagrama_caso_uso.png'))
    print('\n✓ Diagrama exibido com sucesso!')
else:
    print('⚠ Arquivo de imagem não encontrado. Verifique a instalação do PlantUML.')
```

## Descrição dos Casos de Uso

### Gerenciar Acesso
- **UC01: Efetuar Cadastro/Login** - O usuário cria uma conta ou faz login no sistema
- **UC02: Selecionar Personagem/Avatar** - O usuário escolhe seu personagem para jogar

### Operar Partida
- **UC03: Iniciar Partida** - Inicia uma nova partida (single player ou multiplayer)
- **UC04: Criar Sala/Convidar Amigos** - Cria uma sala multiplayer e convida amigos
- **UC05: Realizar Jogada (Turno)** - Executa ações durante o turno do jogo

### Gestão Financeira (Núcleo do Jogo)
- **UC06: Gerenciar Carteira Empresa (70%)** - Controla o saldo destinado ao negócio
- **UC07: Gerenciar Carteira Pessoal (30%)** - Controla recursos para gastos individuais
- **UC08: Formalizar-se como MEI** - Pagamento do DAS para obter bônus e benefícios
- **UC09: Evitar Multa de Gestão** - Evita penalidades por uso indevido de saldo empresarial

### Suporte e Resultados
- **UC10: Consultar Chatbot (Guia do Equilíbrio)** - Interage com assistente de suporte
- **UC11: Visualizar Ranking Final** - Consulta pontuação e classificação

## Relacionamentos
- **Include (→)**: Relacionamento obrigatório entre casos de uso
- **Extend (→)**: Relacionamento opcional, estende funcionalidades
- **Association (→)**: Conexão entre ator e caso de uso
