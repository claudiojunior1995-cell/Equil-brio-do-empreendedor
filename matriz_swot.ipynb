```python
# Criação da Matriz SWOT - Projeto Equilíbrio Empreendedor
```

```python
# PASSO 1: Instalar e importar bibliotecas

!pip install prettytable -q
from prettytable import PrettyTable
```

```python
# PASSO 2: Definir dados da Matriz SWOT

swot_data = {
    'forcas': [
        'Engajamento por conectividade social',
        'Automação total do fluxo',
        'Acessibilidade e Conveniência',
        'Acompanhamento Visual plural'
    ],
    'fraquezas': [
        'Dependência de sincronismo',
        'Complexidade no código do lobby',
        'Barreira técnica do público-alvo'
    ],
    'oportunidades': [
        'Crescimento Orgânico',
        'Adoção escolar e corporativa simplificada',
        'Torneios e gamificação da comunidade'
    ],
    'ameacas': [
        'Custos elevados de servidores',
        'Concorrência com jogos casuais sociais'
    ]
}
```

```python
# PASSO 3: Função para formatar itens com bullets

def formatar_itens(lista):
    """Formata lista de itens com bullets"""
    return '\n'.join([f'• {item}' for item in lista])
```

```python
# PASSO 4: Criar e formatar tabelas

# Tabela 1: Forças e Fraquezas
swot_1 = PrettyTable(['FORÇAS', 'FRAQUEZAS'])
swot_1.align['FORÇAS'] = 'l'
swot_1.align['FRAQUEZAS'] = 'l'
swot_1.max_width['FORÇAS'] = 40
swot_1.max_width['FRAQUEZAS'] = 40
swot_1.add_row([
    formatar_itens(swot_data['forcas']),
    formatar_itens(swot_data['fraquezas'])
])

# Tabela 2: Oportunidades e Ameaças
swot_2 = PrettyTable(['OPORTUNIDADES', 'AMEAÇAS'])
swot_2.align['OPORTUNIDADES'] = 'l'
swot_2.align['AMEAÇAS'] = 'l'
swot_2.max_width['OPORTUNIDADES'] = 40
swot_2.max_width['AMEAÇAS'] = 40
swot_2.add_row([
    formatar_itens(swot_data['oportunidades']),
    formatar_itens(swot_data['ameacas'])
])
```

```python
# PASSO 5: Exibir a Matriz SWOT

print(swot_1)
print()
print(swot_2)
```

```python
# BÔNUS: Exportar dados em formato reutilizável (JSON)

import json

# Salvar em JSON para usar em outros arquivos
with open('swot_data.json', 'w', encoding='utf-8') as f:
    json.dump(swot_data, f, ensure_ascii=False, indent=2)

print('✓ Dados exportados para swot_data.json')
print('\nPara usar em outro arquivo, importe assim:')
print('import json')
print('with open("swot_data.json") as f:')
print('    swot_data = json.load(f)')
```
