```python
# PASSO 2: importar módulo

from prettytable import PrettyTable
```

```python
# PASSO 3: definir nomes das colunas e larguras

minhaSWOT1 = PrettyTable(["FORÇAS", "FRAQUEZAS"])
minhaSWOT2 = PrettyTable(["OPORTUNIDADES", "AMEAÇAS"])

minhaSWOT1._max_width = {"FORÇAS" : 35, "FRAQUEZAS" : 35}
minhaSWOT2._max_width = {"OPORTUNIDADES" : 35, "AMEAÇAS" : 35}
```

```python

# PASSO 4: Adicionando conteúdo adaptado conforme documento

# Linha 1
minhaSWOT1.add_row([
    "• Engajamento por conectividade social\n• Automação total do fluxo\n• Acessibilidade e Conveniência.\n• Acompanhamento Visual plural.",
    "• Dependência de sincronismo\n• Complexidade no código do lobby\n• Barreira técnica do público-alvo"
])

# Linha 2
minhaSWOT2.add_row([
    "• Crescimento Orgânico.\n• Adoção escolar e corporativa simplificada.\n• Torneios e gamificação da comunidade.",
    "• Custos elevados de servidores.\n• Concorrência com jogos casuais sociais."
])
```

```python
# PASSO 5: printar a SWOT completa

print(minhaSWOT1)
print()
print(minhaSWOT2)
```
